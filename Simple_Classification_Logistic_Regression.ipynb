Task 1. Write sigmoid(z) that accepts a real number and returns its sigmoid output. Then write predict_class(z) that calls sigmoid() and returns 1 if the probability ≥ 0.5, else 0.

```python
import math

def sigmoid(z):
  """returns the sigmoid of Z"""
  return 1/(1 + math.exp(-z))


def predict_class(z):
  prob = sigmoid(z)
  if prob >= 0.5:
    return 1
  else:
    return 0
```

Task 2. Write evaluate(y_true, y_pred) that accepts two lists (actual and predicted labels) and returns a dictionary with accuracy, false_positives, and false_negatives.

Task 3. Write display_report(y_true, y_pred, z_values) that calls all above functions and prints:

```python
def evaluate(y_true, y_pred):
  """returns accuracy, FP and FN as a diectionary"""

  accuracy =0
  false_positive =0
  false_negative=0

  for true, pred in zip(y_true,y_pred):
    if true == pred:
      accuracy +=1
    elif true ==0 and pred == 1:
      false_positive +=1
    elif true ==1 and pred == 0:
      false_negative +=1

  return {'accuracy':round(accuracy/len(y_true) * 100,2),  #accuracy is determined by total_accuracy/lenght of original y_true
          'false_positive':false_positive,
          'false_negative':false_negative}


def display_report(y_true, z_values):

  prob = [sigmoid(z) for z in z_values]
  y_pred = [predict_class(z) for z in z_values]
  dict = evaluate(y_true,y_pred)

  print(f"Accuracy: {dict['accuracy']}")
  print(f"False Positives: {dict['false_positive']}")
  print(f"False Negatives: {dict['false_negative']}")
  print(f"Probability :{prob}")
  print(f"Predicted class : {y_pred}")
```
